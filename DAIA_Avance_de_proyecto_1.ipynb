Modulo procesamiento

```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 1 - BakeOps AI

# Fecha: 20-10-2026

################################################################################
## Módulo de procesamiento de consultas
################################################################################
##;

class m_ProcesaConsulta:

  def __init__(self):

    self.stop_words = set(stopwords.words("spanish"))
    # Esta variable va a cargar palabras del español que no aportan información
    # que sea importante para la función de los siguientes módulos,
    # por ejemplo, los artículos el, la, los, o preposiciones.

  def a1_ProcesarConsulta(self,consulta):

    # Ahora inicia la parte del procesamiento del texto
    consulta = consulta.lower() # Bajamos el texto recibido a minúsuclas

    tokens = word_tokenize(consulta) # Los tokens son para separar en palabras
                                    # individuales, aquí todavía incluye signos
                                    # ortográficos, solo se separan.
                                    # Esto divide las palabras en elementos
                                    # llamados token, 1 por cada palabra.

    SetFinal = []   # Inicializamos lista final de palabras procesadas.

    for token in tokens: # Ahora vamos a recorrer cada palabra de la consulta.
      if token.isalnum(): # Este método isalnum, se encarga de filtrar solo texto y
                          # números dejando fuera signos.
        if token not in self.stop_words: # Vemos si el token o la palabra no está en la
                                    # lista de palabras no relevantes
          SetFinal.append(token) # Lo añadimos a la lista

    return SetFinal # La función retorna la lista de la raíz de cada token o palabra.
```

Modulo de analisis de consultas

```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 1 - BakeOps AI

# Fecha: 20-10-2026

################################################################################
## Módulo de análisis de consultas
################################################################################
##;

# Aquí vamos a detectar la intención del usuario, será a de mayor puntaje.

class m_AnalizarConsultas:

  def __init__(self,BaseConocimientos):
    # Se guarda la base de conocimientos de BakeOps
    # para aquí consultar las intenciones y palabras claves
    self.BaseConocimientos = BaseConocimientos

  def a1_AnalizarConsulta(self,consulta):

    # Se inicializan las variables, con los valores por defecto
    # en el caso de que no existan coincidencias, el programa
    # tomaría estos valores.
    MejorIntencion = "desconocida"
    MejorPuntaje = 0

    ConsultaTexto = " ".join(consulta)
    # Aquí vamos a iterar en la base de conocimientos, recorriendo
    # cada intención. Este módulo especificamente busca en las intenciones.
    for Intencion, Datos in self.BaseConocimientos["intenciones"].items():

      PalabrasClave = Datos["palabras_clave"] # Palabras clave de la intención
                                              # de esta iteración
      Puntaje = 0 # Inicializamos puntaje en 0.

      for Palabra in PalabrasClave:  # Recorremos cada palabra clave

        if Palabra in ConsultaTexto:  # Si una palabra clave se encuentra en la
                                 # consulta entramos.

          Puntaje += 1           # Incrementamos el puntaje + 1.

      if Puntaje > MejorPuntaje: # Si el puntaje obtenido es el mayor,
                                   # se actualiza la mejor intención
                                   # y se guarda el puntaje. La indentación
                                   # es importante, porque solo queremos
                                   # que se revisen las puntuaciones al final

        MejorIntencion = Intencion
        MejorPuntaje = Puntaje

    return MejorIntencion, MejorPuntaje # La función devuelve mejor intención y
                                        # mejor puntaje
```

```python
analizador = m_AnalizarConsultas(BaseOriginal)

Intencion, Puntaje = analizador.a1_AnalizarConsulta(["producirá", "hoy"])

print("Intención:", Intencion)
print("Puntaje:", Puntaje)
```

Modulo identificacion de entidades

```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 1 - BakeOps AI

# Fecha: 20-10-2026

################################################################################
## Módulo de identificación de entidades
################################################################################
##;
class m_IdentificarEntidades:

  def __init__(self, BaseConocimientos):

    # Se guarda la base de conocimientos de BakeOps
    # para aquí consultar las entidades que ahí se registran.
    self.BaseConocimientos = BaseConocimientos

  def a1_IdentificarEntidades(self, ConsultaProcesada):

      Entidades ={} # Inicializamos una lista para guardar las entidades que se
                    # encuentren

      ConsultaTexto = " ".join(ConsultaProcesada)

      for Categoria, ListaEntidades in self.BaseConocimientos["entidades"].items():
        # Vamos a recorrer cada categoría que se tenga registrada en la
        # Base de conocimientos

        for Entidad in ListaEntidades:

          # Verificamos si la entidad aparece en la consulta.
          if Entidad in ConsultaTexto:

            # Si existe coincidencia se guarda junto con la categoría
            Entidades[Categoria] = Entidad

      # Regresamos todas las entidades identificadas, para los módulos siguientes
      return Entidades
```

Modulo de generacion de respuestas

```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 1 - BakeOps AI

# Fecha: 21-10-2026

################################################################################
## Módulo de generación de respuestas
################################################################################
##;
# En este módulo se prepara la respuesta al usuario.
# Siguiendo los criterios intención + entidad cuando aplique.

class m_GeneraRespuesta:

 def __init__(self, BaseConocimientos):

    # Se guarda la base de conocimientos de BakeOps
    # para aquí consultar las respuestas que ahí se registran.
    self.BaseConocimientos = BaseConocimientos

 def a1_GenerarRespuesta(self, Intencion, Entidades):

    # Si la intención corresponde con una consulta de receta entra
    if Intencion == "consulta_receta":
      Producto = Entidades.get("productos") # Se obtiene el producto

      if Producto: # Si se encontró el producto, se busca el producto dentro
                   # de la base de conocimientos, de lo contrario, retorna que no
                   # encontró una receta para ese producto
         return self.BaseConocimientos["recetas"].get(Producto,
         "No encontré la receta solicitada.")

    # La lógica para las demás intenciones es la misma, cambiando la entidad
    # y la intención referida.
    if Intencion == "consulta_inventario":

      Ingrediente = Entidades.get("ingredientes")

      if Ingrediente:

          return self.BaseConocimientos["inventarios"].get(
              Ingrediente,"No encontré información de inventario.")

    if Intencion == "produccion_programada":

      Producto = Entidades.get("productos")

      if Producto:

          return self.BaseConocimientos["produccion"].get(
              Producto,"No encontré información de producción.")

    if Intencion == "criterios_calidad":

      Producto = Entidades.get("productos")

      if Producto:

          return self.BaseConocimientos["calidad"].get(
              Producto,"No encontré criterios de calidad.")

    if Intencion == "procedimiento_fabricacion":

      Proceso = Entidades.get("procesos")

      if Proceso:

            return self.BaseConocimientos["detalle_procesos"].get(
                Proceso,"No encontré información del proceso.")

    if Intencion == "consulta_responsable":

      Area = Entidades.get("areas")

      if Area:

            return self.BaseConocimientos["supervisores"].get(
                Area,"No encontré supervisor para esa área.")

    # Si la intención detectada no necesita una entidad en particular,
    # utilizamos alguna de las respuestas generales que ya están definidas con la
    # intención en la base de conocimientos.
    if Intencion in self.BaseConocimientos["intenciones"]:

        return (self.BaseConocimientos["intenciones"][Intencion]["respuesta"])

    # Si en este punto no se encontró ninguna intención coincidente,
    # mostramos la respuesta genérica.
    return ("Lo siento, no entendí tu consulta. Intenta nuevamente.")
```

Modulo historial

```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 1 - BakeOps AI

# Fecha: 21-10-2026

################################################################################
## Módulo de historial
################################################################################
##;

class m_Historial:

  def __init__(self):
    # Por practicidad vamos a guardar el historial en una lista
    self.Historial = []

  def a1_GuardarHistorial(self,Consulta,Intencion,Entidades,Respuesta):

    # Vamos a grabar en un registro, cada resultado de los módulos,
    # con ello generaremos un historial completo desde lo que fue
    # el input del usuario, hasta lo que procesamos con el chat bot y
    # la respuesta que se produjo en esa interacción.

    Registro = {"consulta": Consulta,
                "intencion": Intencion,
                "entidades": Entidades,
                "respuesta": Respuesta}

    # Añadimos el registro a la lista del historial
    self.Historial.append(Registro)

  def a2_ObtenerHistorial(self):
    # esta función solo retorna el historal
    return self.Historial
```

```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 1 - BakeOps AI

# Fecha: 22-10-2026

################################################################################
## Módulo conversacional
################################################################################
##;

# Al ser una arquitectura modular, donde los elementos del chatbot los hice como
# clases, tenemos que crear los objetos de los módulos.

# Obtención de la base de conocimientos
BaseConocimiento = m_BaseConocimiento()

BaseOriginal = BaseConocimiento.a3_ObtenerBase()

# Creación de los módulos del sistema
Procesador = m_ProcesaConsulta() # Procesamiento del texto de consulta
Analizador = m_AnalizarConsultas(BaseOriginal) # Obtención de intención
Identificador = m_IdentificarEntidades(BaseOriginal) # Identificar entidad
Generador = m_GeneraRespuesta(BaseOriginal) # Genera respuesta
Historial = m_Historial() # Guarda historial de conversación para evaluación

print("Conectado con el asistente BakeOps AI")
print("Escriba 'salir' para terminar")

while True: # Ciclo principal, se ejecutá hasta que el usuario escriba salir.

    consulta = input("\nUsuario: ")

    if consulta.lower() == "salir":
        # Mensaje de despedida cuando el usuario escriba salir.
        print("BakeOps AI: Gracias por utilizar el asistente.")
        break

    # Procesamiento de la consulta
    ConsultaProcesada = Procesador.a1_ProcesarConsulta(consulta)

    # Identificación de intención
    Intencion, Puntaje = Analizador.a1_AnalizarConsulta(ConsultaProcesada)

    # Identificación de entidades
    Entidades = Identificador.a1_IdentificarEntidades(ConsultaProcesada)

    # Generación de respuesta
    Respuesta = Generador.a1_GenerarRespuesta(Intencion,Entidades)

    # Registro en historial
    Historial.a1_GuardarHistorial(
        consulta,
        Intencion,
        Entidades,
        Respuesta
    )

  # Herramientas para debugging. (No son necesarias al chatear)
  #  print("Intención detectada:", Intencion)
  #  print("Puntaje:", Puntaje)
  #  print("Entidades:", Entidades)

    # Respuesta del chat bot, identificandolo, y añadiendo la respuesta.
    print("BakeOps AI:", Respuesta)
```

Modulo evaluacion

```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 1 - BakeOps AI

# Fecha: 22-10-2026

################################################################################
## Módulo de evaluación del chatbot
################################################################################
##;
class m_Evaluacion:

  def __init__(self, Historial):

    # Se guarda la referencia al módulo de historial.
    self.Historial = Historial

  def a1_EvaluarChatbot(self):

    Registros = self.Historial.a2_ObtenerHistorial()

    TotalConsultas = len(Registros)

    Desconocidas = 0

    # Se recorren todos los registros almacenados.
    for Registro in Registros:

      if Registro["intencion"] == "desconocida":

        Desconocidas += 1

    Reconocidas = TotalConsultas - Desconocidas

    if TotalConsultas > 0:

      Precision = (Reconocidas / TotalConsultas) * 100

      Fallback = (Desconocidas / TotalConsultas) * 100

    else:

      Precision = 0
      Fallback = 0

    print("\n========== Métricas de evaluación ==========")

    print("Consultas Totales:",TotalConsultas)

    print("Consultas Reconocidas:",Reconocidas)

    print("Consultas No Reconocidas:",Desconocidas)

    print("Precisión:",round(Precision, 2),"%")

    print("Tasa de Fallback:",round(Fallback, 2),"%")
```

```python
Evaluador = m_Evaluacion(Historial)

Evaluador.a1_EvaluarChatbot()
```
