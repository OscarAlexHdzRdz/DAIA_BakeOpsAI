```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 2 - BakeOps AI

# Fecha: 10-11-2026

################################################################################
## Módulo de procesamiento de consultas
################################################################################
##;

class m_ProcesaConsulta:

  def __init__(self):

    self.stop_words = set(stopwords.words("spanish"))
    # Esta variable va a cargar palabras del español que no aportan información
    # que sea importante para la función de los siguientes módulos,
    # por ejemplo, los artículos el, la, los, o preposiciones.

    self.nlp = spacy.load("es_core_news_sm") # Modelo de spaCy para el idioma
                                             # español.
    # Las palabras no relevantes se van a cargar en esta variable.

  def a1_ProcesarConsulta(self,consulta):

    # Ahora inicia la parte del procesamiento del texto
    #*******************************************************
    # 1. Aplica normalización a la consulta textual
    #*******************************************************
    consulta = consulta.lower() # Bajamos el texto recibido a minúsuclas

    #*******************************************************
    # 2. Aplica tokenización
    #*******************************************************
    tokens = word_tokenize(consulta) # Los tokens son para separar en palabras
                                    # individuales, aquí todavía incluye signos
                                    # ortográficos, solo se separan.
                                    # Esto divide las palabras en elementos
                                    # llamados token, 1 por cada palabra.

    #*******************************************************
    # 3. y 4. Elimina stopwords y aplica lematización
    #*******************************************************
    SetFinal = []   # Inicializamos lista final de palabras procesadas.

    for token in tokens: # Ahora vamos a recorrer cada palabra de la consulta.

      if token.isalnum(): # Este método isalnum, se encarga de filtrar solo texto y
                          # números dejando fuera signos.

        if token not in self.stop_words: # Vemos si el token o la palabra no está en la
                                         # lista de palabras no relevantes

          # Aplicamos lematización, para obtener la forma base de la palabra,
          # esto nos ayuda a reducir los plurales, verbos conjugados, etc.
          doc = self.nlp(token)
          lema = doc[0].lemma_

          # print(token, "->", doc[0].lemma_) Esto es para probar la lematización
          # comento para no integrarlo a nuestro código final, ya que no es
          # necesario

          SetFinal.append(lema) # Lo añadimos a la lista

    return SetFinal # La función retorna la lista de la raíz de cada token o palabra.
```

Modulo de analisis de consultas

```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 2 - BakeOps AI

# Fecha: 11-11-2026

################################################################################
## Módulo de análisis de consultas
################################################################################
##;

# Aquí vamos a detectar la intención del usuario, será a de mayor puntaje.

class m_AnalizarConsultas:

  def __init__(self,BaseConocimientos, DatosEntrenamiento):
    # Se guarda la base de conocimientos de BakeOps
    # para aquí consultar las intenciones y palabras claves
    self.BaseConocimientos = BaseConocimientos

    # Dataset para entrenamiento
    self.DatosEntrenamiento = DatosEntrenamiento

    # Representación TF-IDF
    self.Vectorizador = TfidfVectorizer()

    # Clasificador Naive Bayes
    self.Modelo = MultinomialNB()

  def a1_PrepararDataset(self):

    Textos = []        # Hacemos lista para almacenar consultas
    Intenciones = []   # Hacemos lista para almacenar intenciones

    # Aquí vamos a iterar en la base de conocimientos, recorriendo
    # cada intención. Este módulo especificamente busca en las intenciones.
    for ConsultaProcesada,Intencion in self.DatosEntrenamiento:

      # Consulta procesada
      # Se une la consulta para formar la cadena de texo
      TextoProcesado = " ".join(ConsultaProcesada)
      Textos.append(TextoProcesado)   # Se almacena el texto
      Intenciones.append(Intencion)   # Se almacena la intención que le corresponde

    return Textos, Intenciones

  def a2_DividirDataset(self):
      # En este punto ya tenemos textos con etiquetas
      Textos, Intenciones = self.a1_PrepararDataset()

      #Dividimos dataset en entrenamiento y prueba
      self.X_train,self.X_test,self.y_train,self.y_test = train_test_split(Textos,
                                                                           Intenciones,
                                                                           test_size = 0.20,
                                                                           random_state = 42,
                                                                           stratify = Intenciones)

  def a3_VectorizarTFIDF(self):
      # Vocabulario solo con datos de entrenamiento
      self.X_train_vectorizado = self.Vectorizador.fit_transform(self.X_train)
      # Conjunto de pruebas utiliza el vocabulario aprendido en la linea anterior
      self.X_test_vectorizado =  self.Vectorizador.transform(self.X_test)

  def a4_EntrenarClasificador(self):
      # Entrenamiento del modelo Naive Bayes
      self.Modelo.fit(self.X_train_vectorizado,self.y_train)

  def a5_EvaluarClasificador(self):
      # Generación de predicciones del conjunto de prueba
      y_pred = self.Modelo.predict(self.X_test_vectorizado)

      print("*******************************************************")
      print("Reporte de clasificación")
      print("*******************************************************")
      # Aquí obtenemos precisión, recall y F1-score para cada intención
      print(classification_report(self.y_test,y_pred))

      print("*******************************************************")
      print("Matriz de confusión")
      print("*******************************************************")
      # Matriz de confusión (aciertos y errores)
      print(confusion_matrix(self.y_test,y_pred))


  def a6_EntrenarModelo(self):
      # Ejecució completa del proceso de entrenamiento
      # Esto nos muestra claramente el orden de la arquitectura del modulo
      self.a2_DividirDataset() # Division dataset
      self.a3_VectorizarTFIDF() # Vectorización TF-IDF
      self.a4_EntrenarClasificador() # Entrenamiento del clasificador


  def a7_AnalizarConsulta(self, ConsultaProcesada):

      # La consulta ya viene procesada desde el
      # módulo de procesamiento de texto

      TextoProcesado = " ".join(ConsultaProcesada)

      # Transformación de consulta a TD-IDF
      VectorConsulta = self.Vectorizador.transform([TextoProcesado])

      # Intención más probable
      Intencion = self.Modelo.predict(VectorConsulta)[0]

      return Intencion
```

```python
################################################################################
## Prueba completa del módulo de análisis de consultas
################################################################################
##;

# Crear instancia del módulo
modulo_analisis = m_AnalizarConsultas(
    BaseConocimiento,
    DatosEntrenamiento
)

# Entrenar modelo
print("Entrenando clasificador...")
modulo_analisis.a6_EntrenarModelo()

# Evaluar desempeño
print("\nEvaluando clasificador...")
modulo_analisis.a5_EvaluarClasificador()

# Consulta de prueba
Consulta = "Quien es el supervisor"

# Procesamiento de texto
ConsultaProcesada = modulo_procesamiento.a1_ProcesarConsulta(
    Consulta
)

print("\n*******************************************************")
print("Consulta original")
print("*******************************************************")
print(Consulta)

print("\n*******************************************************")
print("Consulta procesada")
print("*******************************************************")
print(ConsultaProcesada)

# Identificación de intención
Intencion = modulo_analisis.a7_AnalizarConsulta(
    ConsultaProcesada
)

print("\n*******************************************************")
print("Intención detectada")
print("*******************************************************")
print(Intencion)
```

```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 1 - BakeOps AI

# Fecha: 20-10-2026

################################################################################
## Módulo de identificación de entidades
################################################################################
##;
class m_IdentificarEntidades:

  def __init__(self, BaseConocimientos):

    # Se guarda la base de conocimientos de BakeOps
    # para aquí consultar las entidades que ahí se registran.
    self.BaseConocimientos = BaseConocimientos

  def a1_IdentificarEntidades(self, ConsultaProcesada):

      Entidades ={} # Inicializamos una lista para guardar las entidades que se
                    # encuentren

      ConsultaTexto = " ".join(ConsultaProcesada)

      for Categoria, ListaEntidades in self.BaseConocimientos["entidades"].items():
        # Vamos a recorrer cada categoría que se tenga registrada en la
        # Base de conocimientos

        for Entidad in ListaEntidades:

          # Verificamos si la entidad aparece en la consulta.
          if Entidad in ConsultaTexto:

            # Si existe coincidencia se guarda junto con la categoría
            Entidades[Categoria] = Entidad

      # Regresamos todas las entidades identificadas, para los módulos siguientes
      return Entidades
```

```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 2 - BakeOps AI

# Fecha: 11-11-2026

################################################################################
## Módulo visual
################################################################################
##;

# Este módulo es responsable de clasificar imágenes
# relacionadas con el control de calidad de productos
# panaderos utilizando ResNet50.

class m_Visual:

  def __init__(self,RutaDataset):

    # Ruta donde se encuentran las imágenes organizadas
    # por categorías.
    self.RutaDataset = RutaDataset

  def a1_CargarDataset(self):

    # Se cargan las imágenes desde las carpetas
    # definidas para cada categoría.
    Dataset = tf.keras.utils.image_dataset_from_directory(

                                                        self.RutaDataset,
                                                        validation_split = 0.20,
                                                        subset = "both",
                                                        seed = 42,
                                                        image_size = (224,224),
                                                        batch_size = 32
                                                        )

    # Se almacena el conjunto de entrenamiento.
    self.TrainDataset = Dataset[0]

    # Se almacena el conjunto de prueba.
    self.TestDataset = Dataset[1]

    # Se guardan los nombres de las categorías.
    self.Clases = self.TrainDataset.class_names

  def a2_PrepararImagenes(self):

    # Se preparan las imágenes de entrenamiento para
    # que puedan ser utilizadas por ResNet50.
    self.TrainDataset = self.TrainDataset.map(
                                              lambda x,y:
                                              (
                                                preprocess_input(x),
                                                y
                                              )
                                             )

    # Se preparan las imágenes de prueba utilizando
    # el mismo procedimiento.
    self.TestDataset = self.TestDataset.map(
                                            lambda x,y:
                                            (
                                              preprocess_input(x),
                                              y
                                            )
                                           )

  def a3_CrearModelo(self):

    # Se carga el modelo ResNet50 previamente
    # entrenado con ImageNet.
    BaseModel = ResNet50(
      weights = "imagenet",
      include_top = False,
      input_shape = (224,224,3)
    )

    # Se congelan las capas previamente entrenadas.
    BaseModel.trainable = False

    # Capa de entrada del modelo.
    Entradas = tf.keras.Input(

      shape = (224,224,3)

    )

    # Las imágenes se envían a ResNet50.
    x = BaseModel(
      Entradas,
      training = False
    )

    # Reduce la salida generada por ResNet50.
    x = tf.keras.layers.GlobalAveragePooling2D()(x)

    # Se utiliza Dropout para reducir sobreajuste.
    x = tf.keras.layers.Dropout(0.30)(x)

    # Capa de salida para generar categoría final.
    Salidas = tf.keras.layers.Dense(
      len(self.Clases),
      activation = "softmax"
    )(x)

    # Construcción del modelo final.
    self.Modelo = tf.keras.Model(
      Entradas,
      Salidas
    )

    # Configuración del entrenamiento.
    self.Modelo.compile(
      optimizer = "adam",
      loss = "sparse_categorical_crossentropy",
      metrics = ["accuracy"]
    )

  def a4_EntrenarModelo(self):

    # Entrenamiento del clasificador visual.
    self.History = self.Modelo.fit(
      self.TrainDataset,
      validation_data = self.TestDataset,
      epochs = 10
    )

  def a5_EvaluarModelo(self):

    # Listas para almacenar valores reales
    # y predicciones generadas.
    y_real = []

    y_pred = []

    # Se recorren imágenes de prueba.
    for Imagenes,Leyendas in self.TestDataset:

      # Se generan predicciones.
      Predicciones = self.Modelo.predict(
        Imagenes,
        verbose = 0
      )

      # Se almacenan clases reales.
      y_real.extend(
        Leyendas.numpy()
      )

      # Se almacenan clases predichas.
      y_pred.extend(
        np.argmax(
          Predicciones,
          axis = 1
        )
      )

    print("*******************************************************")
    print("Accuracy")
    print("*******************************************************")

    # Cálculo de exactitud del modelo.
    Accuracy = accuracy_score(
      y_real,
      y_pred
    )

    print(Accuracy)

    print("*******************************************************")
    print("Reporte de clasificación")
    print("*******************************************************")

    # Se muestran métricas de clasificación
    # para cada categoría.
    print(
      classification_report(
        y_real,
        y_pred,
        target_names = self.Clases
      )
    )

    print("*******************************************************")
    print("Matriz de confusión")
    print("*******************************************************")

    # Se muestran aciertos y errores de clasificación.
    print(
      confusion_matrix(
        y_real,
        y_pred
      )
    )

  def a6_EntrenarClasificadorVisual(self):

    # Ejecución completa del flujo de entrenamiento.
    self.a1_CargarDataset()
    self.a2_PrepararImagenes()
    self.a3_CrearModelo()
    self.a4_EntrenarModelo()

  def a7_ClasificarImagen(self,RutaImagen):

    # Se carga la imagen que desea analizar el usuario.
    Imagen = tf.keras.utils.load_img(
      RutaImagen,
      target_size = (224,224)
    )

    # Conversión de imagen a arreglo numérico.
    Imagen = tf.keras.utils.img_to_array(
      Imagen
    )

    # Preparación de la imagen para ResNet50.
    Imagen = preprocess_input(
      Imagen
    )

    # Se agrega una dimensión adicional para
    # permitir la clasificación.
    Imagen = np.expand_dims(
      Imagen,
      axis = 0
    )

    # Se obtiene la predicción del modelo.
    Prediccion = self.Modelo.predict(
      Imagen,
      verbose = 0
    )

    # Se identifica la posición de la clase
    # con mayor probabilidad.
    IndiceClase = np.argmax(
      Prediccion
    )

    # Obtención del nombre de la clase detectada.
    ClaseDetectada = self.Clases[
      IndiceClase
    ]

    return ClaseDetectada
```

```python
################################################################################
## Prueba completa del módulo visual
################################################################################
##;

# Crear módulo
modulo_visual = m_Visual(
    "dataset_imagenes"
)

# Entrenar
print("Entrenando clasificador visual...")
modulo_visual.a6_EntrenarClasificadorVisual()

# Evaluar
print("\nEvaluando clasificador visual...")
modulo_visual.a5_EvaluarModelo()

# Imagen de prueba
RutaImagen = "pruebas/pan01.jpg"

print("\nImagen analizada:")
print(RutaImagen)

# Clasificación
Resultado = modulo_visual.a7_ClasificarImagen(
    RutaImagen
)

print("\nClasificación visual:")
print(Resultado)
```

```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 2 - BakeOps AI

# Fecha: 11-11-2026

################################################################################
## Módulo de generación de respuestas
################################################################################
##;

# En este módulo se prepara la respuesta al usuario.
# Siguiendo los criterios de intencion + entidad cuando aplique.
# Ahora hay un componente de texto y uno de imagen.

class m_GeneraRespuesta:

  def __init__(self, BaseConocimientos):

    # Se guarda la base de conocimientos de BakeOps
    # para consultar la información necesaria al
    # momento de generar respuestas.
    self.BaseConocimientos = BaseConocimientos

  def a1_GenerarRespuesta(
                          self,
                          Intencion,
                          Entidades,
                          ResultadoVisual = None
                          ):

    # Si existe un resultado proveniente del módulo visual,
    # se genera una respuesta relacionada con la clasificación
    # obtenida por ResNet50.

    if ResultadoVisual:

      if ResultadoVisual == "Normal":

        return (
            "La imagen analizada corresponde a un producto "
            "que cumple con las especificaciones de calidad."
        )

      if ResultadoVisual == "Subhorneado":

        return (
            "La imagen analizada muestra evidencia de una "
            "cocción insuficiente del producto."
        )

      if ResultadoVisual == "Sobrehorneado":

        return (
            "La imagen analizada presenta evidencia de "
            "sobrehorneado o exceso de cocción."
        )

      if ResultadoVisual == "Deforme":

        return (
            "La imagen analizada presenta deformaciones "
            "respecto a las especificaciones definidas."
        )

    # Si la intención corresponde a una consulta de receta,
    # se intenta localizar el producto solicitado.

    if Intencion == "consulta_receta":

      Producto = Entidades.get("productos")

      if Producto:

        return self.BaseConocimientos["recetas"].get(
            Producto,
            "No encontré la receta solicitada."
        )

    # Se consulta información relacionada con inventarios.

    if Intencion == "consulta_inventario":

      Ingrediente = Entidades.get("ingredientes")

      if Ingrediente:

        return self.BaseConocimientos["inventarios"].get(
            Ingrediente,
            "No encontré información de inventario."
        )

    # Se consulta la producción programada para un producto.

    if Intencion == "produccion_programada":

      Producto = Entidades.get("productos")

      if Producto:

        return self.BaseConocimientos["produccion"].get(
            Producto,
            "No encontré información de producción."
        )

    # Se consultan los criterios de calidad asociados
    # a un producto.

    if Intencion == "criterios_calidad":

      Producto = Entidades.get("productos")

      if Producto:

        return self.BaseConocimientos["calidad"].get(
            Producto,
            "No encontré criterios de calidad."
        )

    # Se consulta el procedimiento de fabricación.

    if Intencion == "procedimiento_fabricacion":

      Proceso = Entidades.get("procesos")

      if Proceso:

        return self.BaseConocimientos["detalle_procesos"].get(
            Proceso,
            "No encontré información del proceso."
        )

    # Se consulta el supervisor asociado a un área.

    if Intencion == "supervisor_produccion":

      Area = Entidades.get("areas")

      if Area:

        return self.BaseConocimientos["supervisores"].get(
            Area,
            "No encontré supervisor para esa área."
        )

    # Si la intención detectada tiene una respuesta general
    # definida en la base de conocimientos se utiliza dicha
    # respuesta.

    if Intencion in self.BaseConocimientos["intenciones"]:

      return self.BaseConocimientos["intenciones"][Intencion]["respuesta"]

    # Si no fue posible identificar la consulta se devuelve
    # una respuesta genérica.
    return ("Lo siento, no entendí tu consulta.Intenta nuevamente.")
```

```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 2 - BakeOps AI

# Fecha: 11-11-2026

################################################################################
## Módulo de historial
################################################################################
##;

# Este módulo es para guardar las interacciones realizadas
# durante la ejecución del sistema. Se registran tanto las
# consultas de texto como de imágenes.

class m_Historial:

  def __init__(self):

    # Por practicidad el historial se almacena en una lista.
    self.Historial = []

  def a1_GuardarHistorial(
                          self,
                          Consulta,
                          Intencion,
                          Entidades,
                          Respuesta,
                          Imagen = None,
                          ResultadoVisual = None
                         ):

    # Se crea un registro con toda la información
    # generada durante una interacción.

    Registro = {

      # Consulta de texto realizada por el usuario.
      "consulta": Consulta,

      # Intención identificada por el clasificador.
      "intencion": Intencion,

      # Entidades encontradas dentro de la consulta.
      "entidades": Entidades,

      # Ruta de imagen cuando la interacción
      # corresponde al módulo visual.
      "imagen": Imagen,

      # Resultado generado por el clasificador visual.
      "resultado_visual": ResultadoVisual,

      # Respuesta generada por el asistente.
      "respuesta": Respuesta

    }

    # El registro se agrega al historial general
    # de conversaciones del sistema.
    self.Historial.append(Registro)

  def a2_ObtenerHistorial(self):

    # Esta función regresa todo el historial
    # almacenado hasta el momento.
    return self.Historial
```

```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 2 - BakeOps AI

# Fecha: 11-11-2026
################################################################################
## Módulo conversacional
################################################################################
##;

# Al ser una arquitectura modular, donde los elementos del chatbot
# fueron implementados mediante clases, primero se crean los objetos
# necesarios para operar el sistema.

# Obtención de la base de conocimientos
BaseConocimiento = m_BaseConocimiento()

BaseOriginal = BaseConocimiento.a3_ObtenerBase()

# Creación del módulo de procesamiento
Procesador = m_ProcesaConsulta()

## Preparación del dataset para clasificación de intenciones

# Aquí se procesan las consultas del dataset para
# utilizarlas durante el entrenamiento del clasificador.

DatosEntrenamiento = []

for Consulta, Intencion in DatosEntrenamientoOriginal:

    # Se procesa cada consulta utilizando el módulo NLP.
    ConsultaProcesada = Procesador.a1_ProcesarConsulta(Consulta)

    # Se almacena la consulta procesada junto
    # con la intención correspondiente.
    DatosEntrenamiento.append(
                              (
                                  ConsultaProcesada,
                                  Intencion
                              )
                             )

# Creación de módulos

# Procesamiento de texto
Procesador = m_ProcesaConsulta()

# Clasificador de intenciones
Analizador = m_AnalizarConsultas(BaseOriginal, DatosEntrenamiento)

# Identificación de entidades
Identificador = m_IdentificarEntidades(BaseOriginal)

# Generación de respuestas
Generador = m_GeneraRespuesta(BaseOriginal)

# Historial de conversaciones
Historial = m_Historial()

# Clasificador visual
ModuloVisual = m_Visual("sample_data/dataset_imagenes")


# Entrenamiento de modelos

# Clasificador de intenciones

print("Entrenando clasificador de intenciones...")

# TF-IDF + Naive Bayes
Analizador.a6_EntrenarModelo()

print("Entrenando clasificador visual...")

# ResNet50
ModuloVisual.a6_EntrenarClasificadorVisual()


# Inicio del chatbot

print("\nConectado con el asistente BakeOps AI")
print("Escriba 'salir' para terminar")
print("Escriba 'imagen' para analizar una imagen")

while True:

    # Captura de consulta del usuario.
    Consulta = input("\nUsuario: ")

    # Si el usuario escribe salir,
    # se termina la ejecución del sistema.
    if Consulta.lower() == "salir":

        print(
            "BakeOps AI: Gracias por utilizar el asistente."
        )

        break

    # Si el usuario escribe imagen,
    # se activa el módulo visual.
    if Consulta.lower() == "imagen":

        # Solicita la ruta de la imagen.
        RutaImagen = input(
            "\nIngrese la ruta de la imagen: "
        )

        # Se clasifica la imagen seleccionada.
        ResultadoVisual = ModuloVisual.a7_ClasificarImagen(RutaImagen)

        # Se genera una respuesta utilizando
        # el resultado del clasificador visual.
        Respuesta = Generador.a1_GenerarRespuesta(
                                                  Intencion = None,
                                                  Entidades = {},
                                                  ResultadoVisual = ResultadoVisual
                                                  )

        # Registro de la interacción visual.
        Historial.a1_GuardarHistorial(
                                      Consulta = None,
                                      Intencion = None,
                                      Entidades = {},
                                      Respuesta = Respuesta,
                                      Imagen = RutaImagen,
                                      ResultadoVisual = ResultadoVisual
                                      )

        print("\nBakeOps AI:", Respuesta)

        continue

    # Procesamiento de consulta.
    ConsultaProcesada = Procesador.a1_ProcesarConsulta(Consulta)

    # Identificación de intención.
    Intencion = Analizador.a7_AnalizarConsulta(ConsultaProcesada)

    # Identificación de entidades.
    Entidades = Identificador.a1_IdentificarEntidades(ConsultaProcesada)

    # Generación de respuesta.
    Respuesta = Generador.a1_GenerarRespuesta(Intencion,Entidades)

    # Registro de la interacción.
    Historial.a1_GuardarHistorial(
                                  Consulta,
                                  Intencion,
                                  Entidades,
                                  Respuesta
                              )

    # Herramientas de depuración.
    # print("Consulta procesada:", ConsultaProcesada)
    # print("Intención detectada:", Intencion)
    # print("Entidades detectadas:", Entidades)

    # Respuesta final del chatbot.
    print("BakeOps AI:", Respuesta)
```

Modulo evaluacion

```python
# Nombre: Oscar Alejandro Hernández Rodríguez
# Matrícula: 07304488

# Avance de proyecto 2 - BakeOps AI

# Fecha: 11-11-2026

################################################################################
## Módulo de evaluación del chatbot
################################################################################
##;

class m_Evaluacion:

  def __init__(
      self,
      Historial,
      Analizador,
      ModuloVisual
  ):

    # Referencia al historial de conversaciones
    self.Historial = Historial

    # Referencia al clasificador de intenciones
    self.Analizador = Analizador

    # Referencia al clasificador visual
    self.ModuloVisual = ModuloVisual

  def a1_EvaluarChatbot(self):

    print("\n*******************************************************")
    print("Evaluación del clasificador textual")
    print("*******************************************************")

    self.Analizador.a5_EvaluarClasificador()

    # Evaluación del clasificador visual

    print("\n*******************************************************")
    print("Evaluación del clasificador visual")
    print("*******************************************************")

    self.ModuloVisual.a5_EvaluarModelo()

    # Evaluación conversacional

    print("\n*******************************************************")
    print("Evaluación conversacional")
    print("*******************************************************")

    Registros = self.Historial.a2_ObtenerHistorial()

    TotalConsultas = len(Registros)
    Desconocidas = 0
    ConsultasTexto = 0
    ConsultasImagen = 0

    # Se recorren todos los registros almacenados
    for Registro in Registros:

      if Registro.get("consulta") is not None:

        ConsultasTexto += 1

      if Registro.get("imagen") is not None:

        ConsultasImagen += 1

      if Registro.get("intencion") == "desconocida":

        Desconocidas += 1

    Reconocidas = TotalConsultas - Desconocidas

    if TotalConsultas > 0:

      PrecisionConversacional = (Reconocidas / TotalConsultas) * 100

      Fallback = (Desconocidas / TotalConsultas) * 100

    else:

      PrecisionConversacional = 0
      Fallback = 0

    print("Consultas totales:", TotalConsultas)

    print("Consultas de texto:", ConsultasTexto)

    print("Consultas con imagen:", ConsultasImagen)

    print("Consultas reconocidas:", Reconocidas)

    print("Consultas no reconocidas:", Desconocidas)

    print(
        "Precisión conversacional:",
        round(
            PrecisionConversacional,
            2
        ),
        "%"
    )

    print(
        "Tasa de fallback:",
        round(
            Fallback,
            2
        ),
        "%"
    )
```

```python
Evaluador = m_Evaluacion(Historial, Analizador, ModuloVisual)

Evaluador.a1_EvaluarChatbot()
```
